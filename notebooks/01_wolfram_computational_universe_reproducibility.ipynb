## 2. Elementary Cellular Automata Simulation Engine
We model 1D elementary cellular automata evolving over a discrete space-time horizon.

```python
def simulate_cellular_automaton(rule_number: int, steps: int = 64, width: int = 129):
    """Simulates 1D elementary cellular automaton with a single center seed."""
    rule_bits = [(rule_number >> i) & 1 for i in range(8)]
    grid = np.zeros((steps, width), dtype=np.uint8)
    grid[0, width // 2] = 1

    for t in range(steps - 1):
        for i in range(1, width - 1):
            neighborhood = (grid[t, i - 1] << 2) | (grid[t, i] << 1) | grid[t, i + 1]
            grid[t + 1, i] = rule_bits[neighborhood]
    return grid

# Compute Rule 30, 110, and 90
ca_30 = simulate_cellular_automaton(30, steps=80, width=161)
ca_110 = simulate_cellular_automaton(110, steps=80, width=161)
ca_90 = simulate_cellular_automaton(90, steps=80, width=161)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(ca_30, cmap='inferno', interpolation='nearest')
axes[0].set_title('Rule 30: Chaotic Class 3 Invariant')
axes[0].axis('off')

axes[1].imshow(ca_110, cmap='viridis', interpolation='nearest')
axes[1].set_title('Rule 110: Universal Computation Class 4')
axes[1].axis('off')

axes[2].imshow(ca_90, cmap='magma', interpolation='nearest')
axes[2].set_title('Rule 90: Sierpinski Fractal Lattice')
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

## 3. Causal Multiway Substitution Graphs & Branchial Distance
We model multiway string-substitution rewritings $A \to ABA, B \to BA$ and trace branchial separation over evolution steps.

```python
def multiway_step(tokens, rules):
    next_states = set()
    for token in tokens:
        for i in range(len(token)):
            ch = token[i]
            if ch in rules:
                replacement = token[:i] + rules[ch] + token[i+1:]
                next_states.add(replacement)
    return list(next_states)

rules = {'A': 'ABA', 'B': 'BA'}
current = ['A']
history = [current]

for step in range(4):
    current = multiway_step(current, rules)
    history.append(current)

cardinalities = [len(h) for h in history]
print('Multiway State Foliation Cardinality by Step:', cardinalities)

plt.figure(figsize=(8, 4))
plt.plot(range(len(cardinalities)), cardinalities, 'o-', color='#38bdf8', lw=2)
plt.title('Multiway State-Space Volume Growth')
plt.xlabel('Rewriting Step t')
plt.ylabel('Unique Hypersurface States |Omega(t)|')
plt.grid(True)
plt.show()
```
